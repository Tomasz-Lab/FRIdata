# Demo run

This notebook shows an example way of using FRIdata toolbox.

## Prerequisities

- virtualenv is set up and activated (see README.md)
- FRIdata CLI tool is created
- File with IDs

## 1. Create a dataset

````python
%%bash

export PATH_TO_DATA="<path to data>"
export IDS_FILE_PATH="<path to idx files>"
export DASHBOARD_NAME="testset_inference_over"
export AFDB_PATH="$PATH_TO_DATA/afdb/structures"

source .venv/bin/activate

fridata generate_data \
 -t dataset \
 -d AFDB \
 -c subset \
 --overwrite \
 --version ${DASHBOARD_NAME}  \
 -i ${IDS_FILE_PATH} \
 --input-path ${AFDB_PATH} \
 -e "esm2_t33_650M_UR50D"
````

## 2. Create dashboard

````python
%%bash

# Path to dataset.json generated from 1. task
export PATH_TO_DATA="<path to data>"
export PATH_TO_DATASET_JSON="$PATH_TO_DATA/datasets/AFDB-subset--testset_inference_over/dataset.json"

source .venv/bin/activate

fridata create_dashboard \
 --dataset ${PATH_TO_DATASET_JSON}
````

Generated dashboard is saved to "./reports/AFDB-subset--testset_inference_over.html"

````python
from IPython.display import IFrame

IFrame(src='reports/AFDB-subset--testset_inference_over.html', width=700, height=600)
````
